Bước 1 – 4: Đọc dữ liệu Online Retail.xlsx, làm sạch, xóa hóa đơn tín dụng (C) và thống kê theo quốc gia

```python
print("Họ tên: Đàm Hoàng Lam - MSSV: 6451071039")

import matplotlib.pyplot as plt
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules, fpgrowth

# 1. Đọc dữ liệu từ file Online Retail.xlsx
df = pd.read_excel('Online Retail.xlsx')
display(df.head(5))
df.info()

# 2. Tiền xử lý: cắt bỏ ký tự thừa ở Description, xóa dòng thiếu InvoiceNo và ép kiểu chuỗi
df['Description'] = df['Description'].str.strip()
df.dropna(axis=0, subset=['InvoiceNo'], inplace=True)
df['InvoiceNo'] = df['InvoiceNo'].astype('str')
display(df.head(10))

# 3. Xóa các hóa đơn tín dụng (InvoiceNo có chứa ký tự 'C')
print('Ví dụ hóa đơn tín dụng:')
display(df[df['InvoiceNo'].str.contains('C')].head(5))
df = df[~df['InvoiceNo'].str.contains('C')]

# 4. Thống kê số dòng dữ liệu theo từng quốc gia và vẽ biểu đồ
country_counts = df['Country'].value_counts()
print(country_counts.head(10))

plt.figure(figsize=(10, 5))
country_counts.head(10).plot(kind='bar', color='steelblue')
plt.title('Top 10 quốc gia có nhiều giao dịch nhất')
plt.xlabel('Quốc gia (Country)')
plt.ylabel('Số dòng giao dịch')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Bước 5 – 8: Lọc dữ liệu nước Đức (Germany), chuyển về dạng One-Hot Encoding và xóa cột 'POSTAGE'

```python
print("Họ tên: Đàm Hoàng Lam - MSSV: 6451071039")

# 5 & 6. Lấy hóa đơn từ nước Đức ('Germany'), gom nhóm theo InvoiceNo & Description và chuyển hot encoding
basket = (
    df[df['Country'] == 'Germany']
    .groupby(['InvoiceNo', 'Description'])['Quantity']
    .sum()
    .unstack()
    .reset_index()
    .fillna(0)
    .set_index('InvoiceNo')
)
display(basket.head(5))


# 7. Hàm biến đổi Quantity > 0 thành 1 (One-Hot Encoding)
def encode_units(x):
  if x <= 0:
    return 0
  if x >= 1:
    return 1


basket_sets = basket.map(encode_units)  # Nếu pandas bản cũ dùng .applymap(encode_units)

# 8. Xóa cột 'POSTAGE' (tiền cước vận chuyển)
if 'POSTAGE' in basket_sets.columns:
  basket_sets.drop('POSTAGE', inplace=True, axis=1)

print('Kích thước ma trận giao dịch tại Germany:', basket_sets.shape)
display(basket_sets.head(5))
```

Bước 9 – 11: Tìm tập phổ biến bằng Apriori (min_sup = 5%), sinh luật kết hợp (min_conf = 50%) và vẽ biểu đồ Scatter Plot

```python
print("Họ tên: Đàm Hoàng Lam - MSSV: 6451071039")

# 9. Áp dụng thuật toán Apriori với min_sup = 5% (0.05)
frequent_itemsets_ap = apriori(
    basket_sets, min_support=0.05, use_colnames=True
)
print('Tổng số tập phổ biến tìm được (Apriori):', len(frequent_itemsets_ap))
display(frequent_itemsets_ap.head(10))

# 10. Tạo luật kết hợp với min_conf = 50% (0.5)
rules_ap = association_rules(
    frequent_itemsets_ap, metric='confidence', min_threshold=0.5
)
print('Tổng số luật kết hợp tìm được:', len(rules_ap))

# Chuyển đổi vế trái (antecedents) và vế phải (consequents) từ frozenset sang chuỗi
rules_ap['antecedents_str'] = rules_ap['antecedents'].apply(
    lambda x: ', '.join(list(x))
)
rules_ap['consequents_str'] = rules_ap['consequents'].apply(
    lambda x: ', '.join(list(x))
)

# In ra danh sách các luật tìm được
for idx, row in rules_ap.iterrows():
  print(
      f"Luật {idx+1}: {row['antecedents_str']}  ==>  {row['consequents_str']}"
      f" | Support: {row['support']:.4f} | Confidence: {row['confidence']:.4f}"
      f" | Lift: {row['lift']:.4f}"
  )

# 11. Biểu diễn Support và Confidence của tập luật lên biểu đồ phân tán (Scatter plot)
plt.figure(figsize=(8, 5))
plt.scatter(
    rules_ap['support'],
    rules_ap['confidence'],
    alpha=0.7,
    color='darkorange',
    edgecolors='k',
    s=60,
)
plt.title('Biểu đồ phân tán Support vs Confidence (Germany - Apriori)')
plt.xlabel('Support (Độ hỗ trợ)')
plt.ylabel('Confidence (Độ tin cậy)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

Bước 12: Tìm tập phổ biến và luật kết hợp bằng thuật toán FP-Growth (min_sup = 5%, min_conf = 50%) và so sánh với Apriori

```python
print("Họ tên: Đàm Hoàng Lam - MSSV: 6451071039")

import time

# Đo thời gian và chạy Apriori
t0 = time.time()
freq_ap = apriori(basket_sets, min_support=0.05, use_colnames=True)
rules_ap_cmp = association_rules(freq_ap, metric='confidence', min_threshold=0.5)
time_ap = time.time() - t0

# Đo thời gian và chạy FP-Growth
t0 = time.time()
freq_fp = fpgrowth(basket_sets, min_support=0.05, use_colnames=True)
rules_fp = association_rules(freq_fp, metric='confidence', min_threshold=0.5)
time_fp = time.time() - t0

print(
    f'Apriori   -> Số tập phổ biến: {len(freq_ap)} | Số luật:'
    f' {len(rules_ap_cmp)} | Thời gian: {time_ap:.4f}s'
)
print(
    f'FP-Growth -> Số tập phổ biến: {len(freq_fp)} | Số luật: {len(rules_fp)} |'
    f' Thời gian: {time_fp:.4f}s'
)

display(
    rules_fp[['antecedents', 'consequents', 'support', 'confidence', 'lift']]
)
```
